```python
import polars as pl
from extract_and_transform import extract_and_transform
df = extract_and_transform()
df.head()
```

```python
df.select(
    pl.col(pl.NUMERIC_DTYPES).max()
)
```

```python
df.select(
    pl.col(pl.NUMERIC_DTYPES).min()
)
```

```python
attacking_cols = [
    "Crossing",
    "Finishing",
    "Heading Accuracy",
    "Short Passing",
    "Volleys",
]

df = df.with_columns(
    pl.sum_horizontal(attacking_cols).alias("calculated_total_attacking")
)

invalid = df.filter(
    pl.col("Total Attacking") != pl.col("calculated_total_attacking")
)

print(f"Invalid total attacking rows: {invalid.height}")
```
